```python
import os
import re
import shutil
from collections import defaultdict
from tqdm import tqdm

# --- Configuration ---
#rpath = "/content/drive/MyDrive/ProJx/UTube"
dirname = 'FtingOCT'
wdir = os.path.join(rpath, dirname)
odir = os.path.join(rpath, 'FtingOCT25MP3')
os.makedirs(odir, exist_ok=True)

def clean_filename(name):
    base, ext = os.path.splitext(name)
    # Keep only alphabetic characters and spaces
    cleaned = re.sub(r'[^a-zA-Z\s]', ' ', base)
    # Normalize whitespace
    cleaned = re.sub(r'\s+', ' ', cleaned).strip()
    # Replace spaces with underscores and lowercase
    cleaned = cleaned.replace(' ', '_').lower()
    return cleaned + ext.lower()

# Get list of files
all_items = os.listdir(wdir)
# Filter only files (not directories)
filenames = [f for f in all_items if os.path.isfile(os.path.join(wdir, f))]

# Initialize counters
stats = {
    'total_processed': 0,
    'copied_success': 0,
    'failed': 0,
    'duplicates_handled': 0,
    'skipped_non_file': len(all_items) - len(filenames)  # dirs or special files
}

# Track cleaned names to detect duplicates
name_count = defaultdict(int)
failed_files = []

# Process with progress bar
for original in tqdm(filenames, desc="Cleaning & Copying Files"):
    stats['total_processed'] += 1
    src_path = os.path.join(wdir, original)
    dst_path = None

    try:
        cleaned = clean_filename(original)
        name_count[cleaned] += 1

        if name_count[cleaned] > 1:
            base, ext = os.path.splitext(cleaned)
            cleaned = f"{base}_{name_count[cleaned] - 1}{ext}"
            stats['duplicates_handled'] += 1

        dst_path = os.path.join(odir, cleaned)

        # Copy file
        shutil.copy2(src_path, dst_path)
        stats['copied_success'] += 1

    except Exception as e:
        stats['failed'] += 1
        failed_files.append((original, str(e)))
        continue

# --- Final Report ---
print("\n" + "="*60)
print("📁 FILE CLEANING & COPY REPORT")
print("="*60)
print(f"Total items in source directory: {len(all_items)}")
print(f"Files processed (non-directories): {stats['total_processed']}")
print(f"✅ Successfully copied:           {stats['copied_success']}")
print(f"⚠️  Duplicates renamed:           {stats['duplicates_handled']}")
print(f"📂 Skipped (dirs/special):       {stats['skipped_non_file']}")
print(f"❌ Failed to copy:                {stats['failed']}")

if failed_files:
    print("\n⚠️  Failed files:")
    for orig, err in failed_files[:10]:  # show first 10 errors
        print(f"  - {orig} → {err}")
    if len(failed_files) > 10:
        print(f"  ... and {len(failed_files) - 10} more.")

print("\n✅ Done!")
```

```python
import os
import re
import sys
import argparse
from pathlib import Path
from tqdm import tqdm
import yt_dlp

HISTORY_FILE = "download_history.txt"

def sanitize_filename(name):
    if not name:
        name = "untitled"
    name = re.sub(r'[^\w\s.-]', '', str(name))
    name = re.sub(r'\s+', '_', name)
    name = re.sub(r'[-_]+', '_', name)
    name = name.strip('_')
    if not name:
        name = "untitled"
    return name

def log_history(playlist_url, output_dir):
    with open(HISTORY_FILE, 'a', encoding='utf-8') as f:
        f.write(f"{playlist_url} | {os.path.abspath(output_dir)}\n")

def utube_playlist_download(playlist_url, output_dir, dry_run=False, max_dry=3):
    output_path = Path(output_dir)
    output_path.mkdir(parents=True, exist_ok=True)

    # Store failed downloads
    failed_downloads = []

    # Progress bar placeholder
    pbar = None

    def progress_hook(d):
        nonlocal pbar
        if d['status'] == 'downloading':
            if pbar is None:
                total = d.get('total_bytes') or d.get('total_bytes_estimate') or 1
                pbar = tqdm(total=total, unit='B', unit_scale=True, desc="Downloading", leave=False)
            if pbar:
                downloaded = d.get('downloaded_bytes', 0)
                pbar.n = min(downloaded, pbar.total)
                pbar.refresh()
        elif d['status'] == 'finished':
            if pbar:
                pbar.close()
                pbar = None

    # Custom outtmpl using sanitized title
    def outtmpl_hook(d):
        title = d.get('title') or 'unknown'
        clean_title = sanitize_filename(title)
        return str(output_path / f"{clean_title}.%(ext)s")

    ydl_opts = {
        'format': 'bestaudio/best',
        'postprocessors': [{
            'key': 'FFmpegExtractAudio',
            'preferredcodec': 'mp3',
            'preferredquality': '192',
        }],
        'outtmpl': outtmpl_hook,
        'noplaylist': False,
        'ignoreerrors': True,
        'quiet': True,
        'no_warnings': True,
        'progress_hooks': [progress_hook],
        'ffmpeg_location': None,  # Set if needed
    }

    with yt_dlp.YoutubeDL(ydl_opts) as ydl:
        try:
            print("Fetching playlist info...")
            info = ydl.extract_info(playlist_url, download=False)
            if not info or 'entries' not in info:
                print("❌ No playlist data found.")
                return

            entries = [e for e in info['entries'] if e]
            if dry_run:
                print(f"🧪 Dry run: Processing only first {min(max_dry, len(entries))} items.")
                entries = entries[:max_dry]

            print(f"📥 Starting download of {len(entries)} items to: {output_path}")
            for i, entry in enumerate(entries, 1):
                title = entry.get('title', 'unknown')
                url = entry.get('webpage_url', 'unknown')
                clean_title = sanitize_filename(title)
                expected_file = output_path / f"{clean_title}.mp3"

                # Handle duplicates
                counter = 1
                final_path = expected_file
                while final_path.exists():
                    stem = expected_file.stem
                    final_path = output_path / f"{stem}_{counter}.mp3"
                    counter += 1

                if final_path.exists():
                    print(f"✅ [{i}/{len(entries)}] Already exists: {final_path.name}")
                    continue

                try:
                    print(f"🎵 [{i}/{len(entries)}] {title}")
                    ydl.download([url])
                    # yt-dlp already uses our sanitized outtmpl, so file should be correct
                    # But in rare cases, double-check and rename if needed
                    if not final_path.exists():
                        # Fallback: look for any .mp3 with similar name
                        candidates = list(output_path.glob(f"{clean_title}*.mp3"))
                        if candidates:
                            os.rename(candidates[0], final_path)
                        else:
                            raise FileNotFoundError(f"Expected MP3 not found for {title}")
                except Exception as e:
                    print(f"❌ [{i}/{len(entries)}] Failed: {title} – {str(e)[:100]}")
                    failed_downloads.append((title, url))

        except Exception as e:
            print(f"💥 Playlist processing failed: {e}")
            return

    # Final report
    print("\n" + "="*60)
    if failed_downloads:
        print(f"⚠️  {len(failed_downloads)} download(s) failed.")
        failure_log = output_path / "failed_downloads.txt"
        with open(failure_log, 'w', encoding='utf-8') as f:
            for title, url in failed_downloads:
                f.write(f"{title} | {url}\n")
        print(f"📝 Failure log saved to: {failure_log}")
    else:
        print("✅ All downloads completed successfully!")

    # Log to history
    log_history(playlist_url, output_dir)

def main():
    parser = argparse.ArgumentParser(description="Download YouTube playlist as MP3.")
    parser.add_argument("url", nargs='?', help="YouTube playlist URL")
    parser.add_argument("-o", "--output", default="downloads", help="Output directory (default: downloads)")
    parser.add_argument("--dry-run", action="store_true", help="Download only first few items for testing")
    parser.add_argument("--max-dry", type=int, default=3, help="Number of items in dry run (default: 3)")

    args = parser.parse_args()

    playlist_url = args.url
    if not playlist_url:
        playlist_url = input("Enter YouTube playlist URL: ").strip()
        if not playlist_url:
            print("❌ No URL provided. Exiting.")
            sys.exit(1)

    utube_playlist_download(
        playlist_url=playlist_url,
        output_dir=args.output,
        dry_run=args.dry_run,
        max_dry=args.max_dry
    )

if __name__ == "__main__":
    main()
```
