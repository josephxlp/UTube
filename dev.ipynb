##### 04 - audio merge

```python
from pydub import AudioSegment

def merge_audio(audio1_path, audio2_path, output_path):
    audio1 = AudioSegment.from_file(audio1_path)
    audio2 = AudioSegment.from_file(audio2_path)

    merged = audio1.overlay(audio2)

    # Export the merged audio to a file
    merged.export(output_path, format='mp3')  # Change format if needed

# Replace the file paths with your audio files
audio_file1 = 'a1.mp3'  # Replace with the first audio file path
audio_file2 = 'a2.mp3'  # Replace with the second audio file path
output_audio = 'two_merged.mp3'  # Replace with the output file path

# Call the function to merge the audio files
merge_audio(audio_file1, audio_file2, output_audio)
```

##### 05 audio concatinate

```python
from pydub import AudioSegment

def concatenate_audio(audio1_path, audio2_path, output_path):
    audio1 = AudioSegment.from_file(audio1_path)
    audio2 = AudioSegment.from_file(audio2_path)

    # Concatenate the two audio segments
    concatenated_audio = audio1 + audio2

    # Export the concatenated audio to a file
    concatenated_audio.export(output_path, format='mp3')  # Change format if needed

# Replace the file paths with your audio files
audio_file1 = 'a1.mp3' #12 sec # Replace with the first audio file path
audio_file2 = 'a2.mp3'  # Replace with the second audio file path
output_audio = 'added_audio.mp3'  # Replace with the output file path

# Call the function to concatenate the audio files
concatenate_audio(audio_file1, audio_file2, output_audio)

```

##### 06 audio delay

```python
from pydub import AudioSegment

# Load the audio file
audio = AudioSegment.from_file("a2.mp3")

# Set the delay in milliseconds
delay_ms = 7000

# Create a delayed version of the audio
delayed_audio = audio[:delay_ms]

# Apply an echo effect by combining the original audio with the delayed version
echoed_audio = audio.overlay(delayed_audio, loop=True)

# Export the echoed audio to a file
echoed_audio.export("output_audio_with_echo.mp3", format="mp3")
```

##### 07-audio add reverb

```python
from pydub import AudioSegment

def add_reverb(audio, delay_ms=100, decay=0.4):
    repetitions = 3
    reverb = AudioSegment.silent(duration=0)

    for i in range(repetitions):
        delayed_audio = audio[:delay_ms]
        decayed_audio = delayed_audio - ((-20 * i) * (i + 1))  # Adjust the attenuation factor
        reverb = reverb.overlay(decayed_audio)

    return audio.overlay(reverb.apply_gain_stereo(decay))

# Load the audio file
audio = AudioSegment.from_file("faisal.m4a")

# Apply reverb effect
reverbed_audio = add_reverb(audio)

# Export the reverbed audio to a file
reverbed_audio.export("output_audio_with_reverb.mp3", format="mp3")

```

##### 08 - audion denoise/clean

```python
from pydub import AudioSegment
from pydub.playback import play
import numpy as np

# Load audio file
audio_file = "add.mp3"#"abbas.wav"
audio = AudioSegment.from_file(audio_file)

# Convert to numpy array
y = np.array(audio.get_array_of_samples())

# Perform simple spectral subtraction for noise reduction
noise_level = np.mean(np.abs(y))
y_denoised = y - 0.5 * noise_level  # Adjust the factor as needed

# Ensure values are within the valid range for 16-bit PCM audio (-32768 to 32767)
y_denoised = np.int16(np.clip(y_denoised, -32768, 32767))

# Convert numpy array back to pydub audio segment
reduced_audio = AudioSegment(
    y_denoised.tobytes(),
    frame_rate=audio.frame_rate,
    sample_width=y_denoised.dtype.itemsize,
    channels=audio.channels
)

# Save the cleaned audio to a file
output_file = "Abbas_output_audio.wav"
reduced_audio.export(output_file, format="wav")

print(f"Cleaned audio saved as '{output_file}'")
```

##### 15- speech rec

```python
import speech_recognition as sr

# Initialize recognizer class (for recognizing the speech)
recognizer = sr.Recognizer()

# Load your audio file
audio_file = "Abbas_output_audio.wav"#"abbas.wav"

# Use the audio file as the audio source
with sr.AudioFile(audio_file) as source:
    # Listen for the data (load audio to memory)
    audio_data = recognizer.record(source)
    # Recognize (convert from speech to text)
    try:
        text = recognizer.recognize_google(audio_data)
        print("Recognized text:")
        print(text)
    except sr.UnknownValueError:
        print("Google Web Speech API could not understand the audio")
    except sr.RequestError as e:
        print(f"Could not request results from Google Web Speech API; {e}")
```

```python
import librosa
import numpy as np

# Path to the audio file
audio_file = 'faisal.m4a'

# Load the audio file
audio_data, sample_rate = librosa.load(audio_file)

# Extract features
# 1. Mel-frequency cepstral coefficients (MFCC)
mfccs = librosa.feature.mfcc(y=audio_data, sr=sample_rate, n_mfcc=13)

# 2. Chroma feature
chroma = librosa.feature.chroma_stft(y=audio_data, sr=sample_rate)

# 3. Spectral Centroid
spectral_centroid = librosa.feature.spectral_centroid(y=audio_data, sr=sample_rate)

# 4. Spectral Contrast
spectral_contrast = librosa.feature.spectral_contrast(y=audio_data, sr=sample_rate)

# 5. Zero Crossing Rate
zero_crossing_rate = librosa.feature.zero_crossing_rate(y=audio_data)

# 6. Spectral Bandwidth
spectral_bandwidth = librosa.feature.spectral_bandwidth(y=audio_data, sr=sample_rate)

# Display the shape of extracted features
print(f"MFCC shape: {mfccs.shape}")
print(f"Chroma shape: {chroma.shape}")
print(f"Spectral Centroid shape: {spectral_centroid.shape}")
print(f"Spectral Contrast shape: {spectral_contrast.shape}")
print(f"Zero Crossing Rate shape: {zero_crossing_rate.shape}")
print(f"Spectral Bandwidth shape: {spectral_bandwidth.shape}")

```

```python
import eyed3

def tag_audio(file_path, artist, album, genre):
    audiofile = eyed3.load(file_path)
    audiofile.tag.artist = artist
    audiofile.tag.album = album
    audiofile.tag.genre = genre
    audiofile.tag.save()

# Example usage
tag_audio("your_audio_file.mp3", "Artist Name", "Album Title", "Genre")
```

```python
from pydub import AudioSegment

def check_loudness(file_path):
    audio = AudioSegment.from_file(file_path)
    loudness = audio.dBFS  # Decibels relative to full scale
    return loudness

# Example usage
print(check_loudness("audio_file.mp3"))
```
